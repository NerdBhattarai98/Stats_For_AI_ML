<h1>Probability Basics</h1>

- Sample Space: A set of all possible outcome
- Event: Paricular Outcome or Group of outcomes
- Probability : A value of occurance between 0 and 1

Example: Roll a fair six-sided die.

Sample Space = {1,2,3,4,5,6}
Event A = Rolling an even number {2,4,6}
Probability of rolling even number = (3/6)

```python

# 1. Load your dataset
df = pd.read_csv(r"C:\Users\Anbit Bhattarai\Desktop\Stats_For_AI_ML\football_practice.csv")

# 2. Create a column for home wins (True if home score > away score)
df['home_win'] = df['home_score'] > df['away_score']

# 3. Shrink the universe: Filter *only* for Friendlies
friendlies = df[df['tournament'] == 'Friendly']

# 4. Find the probability by taking the mean of home_win in that subset
prob_win_given_friendly = friendlies['home_win'].mean()

print(f"Probability of Home Win given Friendly: {prob_win_given_friendly:.4f}")
```

```python
# What is the probability that the away team wins, given that the match is played in Brazil?

played_in_brazil = df[df["country"] == "Brazil"]
print(played_in_brazil.count())

away_win = df[df["home_score"] < df["away_score"]]
print(away_win)

probability = (played_in_brazil["away_score"] > played_in_brazil["home_score"]).mean()
print(probability)
```

```python
world_cup = df[df["tournament"] == "FIFA World Cup"]

# Add home and away scores together right inside the condition
prob = (world_cup["home_score"] + world_cup["away_score"] > 3).mean()
print(prob)
```

```python
# What is the probability that the home team scores at least 1 goal, given that the match is a "Friendly"?

friendlies = df[df["tournament"] == "Friendly"]

probability = (friendlies["home_score"] >= 1).mean()
print(probability)
```

```python
# What is the probability that a match is played at a neutral venue, given that the home team is "Brazil"?

Braziol = df[df["home_team"] == "Brazil"]

# print(Braziol.count())

probability = (Braziol["neutral"] == True).mean()
print(probability)
```

```python
# What is the probability that the home team wins, given that the tournament is "UEFA Euro"?
print(df["tournament"].unique())
Euro = df[df["tournament"] == "UEFA Euro qualification"]

probability  = (Euro["home_score"] > Euro["away_score"]).mean()
print(probability)
```

```python
# What is the probability that the away team wins, given that it is not a neutral venue?

neutral_venue = df[df["neutral"] == False]

probability = (neutral_venue["away_score"] > neutral_venue["home_score"]).mean()
print((neutral_venue["away_score"] > neutral_venue["home_score"]).sum())
print(probability)
```

```python
# What is the probability that a match ends in a draw, given that the tournament is "Copa América"?

tournaments = df[df["tournament"] == "Copa América"]

probability = (tournaments["home_score"] == tournaments["away_score"])
print(probability)
```

```python
# What is the probability that a match is played in "England", given that the away team is "England"?

England = df[df["away_team"] == "England"]

probability = (England["country"] == "England").mean()
print(probability)
```

```python
# What is the probability that total goals are 0, given that the match is a "Friendly"?

matches = df[df["tournament"] == "Friendly"]

probability = (matches["home_score"] + matches["away_score"] == 0).mean()
print(probability)
```

```python
# What is the probability that the home team scores 3 or more goals, given that the match is played at a neutral venue?

matches = df[df["neutral"] == True]

probability = (matches["home_score"] > 3).mean()
print(probability)
```

```python
# What is the probability that the home team wins, given that the tournament is a "Friendly" AND the venue is neutral?

friedly_neutral = df[(df["tournament"] == "Friendly") & (df["neutral"] == False)]

probability = (friedly_neutral["home_score"]>friedly_neutral["away_score"]).mean()
print(probability) 
```

```python
# Given that the tournament is a "Friendly", what is the probability that the home team wins AND total goals are greater than 3?

friendlies = df[df["tournament"] == "Friendly"]

probability = (((friendlies["home_score"] > friendlies["away_score"])) & ((friendlies["home_score"] + friendlies["away_score"]).ge(3))).mean()

print(probability)
```

- Mutual Exclusivity:
    - Two events are mutually exclusive if they cannot happen at the same time. If Event A happens, Event B is strictly impossible.
    - The probability of both happening together is zero ($P(A \text{ and } B) = 0$).
    - Football Example: Can a match result be both a home win and a home loss at the exact same time?

```python
impossible_match = df[(df["home_score"] > df["away_score"]) & (df["home_score"] < df["away_score"])]
print(len(impossible_match))  
```

* Two events are independent if knowing that Event A happened does not change the probability of Event B happening.
    - $P(A \mid B) = P(A)$. (The conditional probability equals the normal marginal probability)
    - Does knowing that a match is a "Friendly" change the baseline probability that a home team wins globally?

```python
# 1. Baseline probability of a home win overall
global_home_win = (df["home_score"] > df["away_score"]).mean()

# 2. Conditional probability of a home win, given that it's a Friendly
friendly_subset = df[df["tournament"] == "Friendly"]
conditional_home_win = (friendly_subset["home_score"] > friendly_subset["away_score"]).mean()

print("Global Home Win Probability:", global_home_win)
print("Friendly Home Win Probability:", conditional_home_win)
# If these numbers are very different, tournament type and home wins are DEPENDENT!
```
